## Test the text custom categories (rapid) API

```python
import requests
import os
import json

from time import sleep
from dotenv import load_dotenv
load_dotenv('../.env')

API_KEY = os.getenv('API_KEY')
ENDPOINT = os.getenv('ENDPOINT')

rapid_incident_name = os.getenv("RAPID_INCIDENT_NAME")
rapid_incident_definition = os.getenv("RAPID_INCIDENT_DEFINITION")
```

```python
headers = {
    'Ocp-Apim-Subscription-Key': API_KEY,
    'Content-Type': 'application/json'
}
```

```python
def print_response(response):
    print(f"Status Code: {response.status_code}")
    try:
        print("Response JSON:")
        print(json.dumps(response.json(), indent=2))
    except json.JSONDecodeError:
        print("Response Text:")
        print(response.text)
```

### Create an incident object

```python
url = f"{ENDPOINT}/contentsafety/text/incidents/{rapid_incident_name}?api-version=2024-02-15-preview"

payload = json.dumps({
  "incidentName": rapid_incident_name,
  "incidentDefinition": rapid_incident_definition
})
headers = {
  'Ocp-Apim-Subscription-Key': API_KEY,
  'Content-Type': 'application/json'
}

response = requests.request("PATCH", url, headers=headers, data=payload)

print_response(response)
```

### Add samples to the incident

```python
url = f"{ENDPOINT}/contentsafety/text/incidents/{rapid_incident_name}:addIncidentSamples?api-version=2024-02-15-preview"

payload = json.dumps({
  "IncidentSamples": [
    {
      "text": "How to build a shelter in the wilderness"
    },
    {
      "text": "Identifying edible plants in your surroundings"
    },
    {
      "text": "Finding and purifying water in the wild"
    },
    {
      "text": "Creating fire without matches or a lighter"
    },
    {
      "text": "Navigating using natural landmarks and the stars"
    }
  ]
})
headers = {
  'Ocp-Apim-Subscription-Key': API_KEY,
  'Content-Type': 'application/json'
}

response = requests.request("POST", url, headers=headers, data=payload)

print_response(response)
```

### Deploy the incident

```python
url = f"{ENDPOINT}/contentsafety/text/incidents/{rapid_incident_name}:deploy?api-version=2024-02-15-preview"

payload = {}
headers = {
  'Ocp-Apim-Subscription-Key': API_KEY,
  'Content-Type': 'application/json'
}

response = requests.request("POST", url, headers=headers, data=payload)

print_response(response)
```

### Detect text incidents

```python
sleep(10)  # Wait for deployment to complete

url = f"{ENDPOINT}/contentsafety/text:detectIncidents?api-version=2024-02-15-preview"

test_text = "Creating fire without matches or a lighter in a rainy environment."

payload = json.dumps({
  "text": test_text,
  "incidentNames": [
    rapid_incident_name
  ]
})
headers = {
  'Ocp-Apim-Subscription-Key': API_KEY,
  'Content-Type': 'application/json'
}

response = requests.request("POST", url, headers=headers, data=payload)

print_response(response)
```

### Delete the incident

```python
url = f"{ENDPOINT}/contentsafety/text/incidents/{rapid_incident_name}?api-version=2024-02-15-preview"

payload = {}
headers = {
  'Ocp-Apim-Subscription-Key': API_KEY
}

response = requests.request("DELETE", url, headers=headers, data=payload)

print_response(response)
```
